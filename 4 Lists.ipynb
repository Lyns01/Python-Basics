**Iterate through a list**

```python
#Loop thru
letters = ['a', 'b', 'c']
for letter in letters:
   print(letter)
```

```python
#Loop thru with Index
letters = ['a', 'b', 'c']
for index, letter in enumerate(letters):
   print(index, letter)
```

```python
#Loop thru with custom index
letters = ['a', 'b', 'c']
for index, letter in enumerate(letters,2):
   print(index, letter)
```

**Iterator and Iterable**

```python
#List is iterable - Use for loop to iterate thru an iterable
lista = [1,2,3,4,5]
for item in lista:
    print(item)
```

```python
#Convert a list to an Iterator
iterB =  iter(lista)
iterB
```

```python
#Iterate thru the iterator - as u keep executing this command, it moves from index 0 to last one after which it throws error
next(iterB)
```

```python
#To handle this error try exception handling - Similar to using list and a for loop
lista = [1,2,3,4,5]
iterB =  iter(lista)
try :
    while true:
        next(iterB)
except :
    print("Last Iteration reached")
```

**List Comprehension**

```python
#Nomal way
lista = [1,2,3,4,5]
listb = []
for item in lista:
    if item %2 == 0:
        listb.append(item*item)
listb
```

```python
#Syntax [WhatToDo ForLoop IfCondition]
#With Condition
lista = [1,2,3,4,5]
listb = [x*x for x in lista if x%2==0]
listb
```

```python
#With Condition - If Else
lista = [1,2,3,4,5]
listb = ['Even' if x%2==0 else 'Odd' for x in lista ]
listb
```

```python
#Without Condition
lista = [1,2,3,4,5]
listb = [x*x for x in lista]
listb
```

```python
#For loop in a line
# Syntax a = [expression for item in iterable if condition == True]
a = [a + 2 for a in range(1,5)]
a
```

**Copy**

```python
#Copy by reference - b still points to the same array as a
a = [1,2,3]
b = a
print('Initially')
print(a)
print(b)
a[0] = 8
print('After changing a value')
print(a)
print(b)
```

```python
#Copy the list to create a new list
a = [1,2,3]
b = a.copy()
print('Initially')
print(a)
print(b)
a[0] = 8
print('After changing a value')
print(a)
print(b)
```
