```python
#Simple example - without any class attributes - which will allow the user to add any number of attributes
class Car:
    pass

Car1 = Car()

Car1.window = 3
Car1.gate = 2
print(Car1.gate)
```

```python
#Class with attributes - __init__ is called the constructor
class Car:
    def __init__(self,carwindow,cardoor) : #Dummy variables like the ones we use for functions
        self.window = carwindow
        self.door = cardoor
        
    def self_driving(self,a): #All functions you define inside a class should have "self" as the first argument
        if a > 5:
            print("This is a self driving car")
        else:
            print("This is not a self driving car")

Car1 = Car() #Throws error coz it expects arguments
```

```python
Car1 = Car(4,5) #By Default the constructor function is called
Car1.window
```

```python
#To call other class functions - object.function()
Car1.self_driving(7)
```

In the above examples,

Car is a class

Car1 is an object that belong the class

Window is the attribute of a class

self_driving() is a method of the class

**Inheritance**

```python
#Base Class
class Car():
    def __init__(self,a,b):
        self.window = a
        self.door = b
    def self_driving(self,num1): 
        if num1 > 5:
            print("This is a self driving car")
        else:
            print("This is not a self driving car")
```

```python
#Inherited class
class Audi(Car):
    def __init__(self,a,b,c):
        super().__init__(a,b)
        self.ai = c
    def enable_ai(self,bool2):
        if bool:
            print("AI Enabled")
        else:
            print("AI Disabled")
```

```python
AudiCar1 = Audi(4,4,True)
#Accessing the attributes
print(AudiCar1.window)
print(AudiCar1.door)
print(AudiCar1.ai)
```

```python
#Accessing Functions 
#Both base and inherited functions can be called thru the inherited class objetc itself
#If both functions have same name then the inherited class function will be executed
AudiCar1.self_driving(6)
```

```python
AudiCar1.enable_ai(True)
```

```python
#To see all the attributes and functions of a class
dir(AudiCar1) #If it starts with __ they are default class attributes and functions, while the rest are user defined
```
