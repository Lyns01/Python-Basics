**Basic Operations**

```python
#Length of a string
name = 'Lynda'
len(name)
```

```python
#Reverse a string - Using built in function
a = "Albert"
a[::-1] #start,end - you know, adding one more param -1 means reverse the string
```

```python
#Direct Method
a = 'Albert'

for i,value in enumerate(a):
    b = len(a)-i-1
    print(a[b])
```

```python
#Split - my_string.split() will split wrt spaces
#Split - my_string.split(',') will split wrt to ","
my_string = 'Do you know the muffin man?'
my_string.split()
```

```python
#Partition - always breaks the string into two part ( + the partinior . or , or whatever it is)
my_string = 'https://www.google.com/'
my_string.partition('.')
```

Difference between split and partition is that partition will look for the first . (dot) and split it into two equal halves. 
Other . (dot) wont be conisdered.

```python
#Join - Join iterable_of_strings with separator_string
separator_string = ' '
iterable_of_strings = ['Happy', 'birthday', 'to', 'you', 'Darling']
separator_string.join(iterable_of_strings)
```

```python
separator_string = ','
iterable_of_strings = ['Happy', 'birthday', 'to', 'you', 'Darling']
separator_string.join(iterable_of_strings)
```

**Print Operations**

**Strip Function**

```python
#Remove the leading and trailing characters (wont remove if they are in the middle)
txt = ",,,,,rrttgg....banana.....rrr"
x = txt.strip(",.grt")
print(x)
```

```python
#Remove the leading and trailing backspaces (at the end or begining only)
txt = "    ban  ana    "
x = txt.strip()
print(x)
```

**Boolean Functions**

```python
#Returns boolean variables 
name = 'Lyn10'

#Either alpha or number or both but not symbols
print('Does it have alphabets and numbers : {}'.format(name.isalnum())) 

print('Does it have only alphabets : {}'.format(name.isalpha()))
print('Does it have only numbers : {}'.format(name.isdigit()))
print('Is it a title : {}'.format(name.istitle())) #Starts with Capital Letter
```

```python
name ='sss'
print('Does it have only Upper Case letters :{}'.format(name.isupper()))
print('Does it have only Lower Case letters :{}'.format(name.islower()))
```

```python
name ='Link'
print('Does it end with k :{}'.format(name.endswith('k'))) 
print('Does it start with l :{}'.format(name.startswith('L'))) #case sensitive
```
