```python
#Types of Arguments 
def func(a,b=30): #a is positional arguments and b is keyword argument
    print(a)
    print(b)
func(20)
```

```python
#*args - positional arguments and **kwargs - keyword arguments
def func(*args,**kwargs):
    print(args) #tuple
    print(kwargs) #dictionary
func('Hai','Hello',a='sheet',b=1)
```

```python
#Passing it as a tuple and a dictionary
def func(*args,**kwargs):
    print(args)
    print(kwargs)
    
listargs = ('Hai','Hello')
dictkwargs = {'a':'sheet','b':1}

func(listargs,dictkwargs) #This has passed the two variables as positional arguments to the function
```

```python
#To pass a list / dict to as args and kwargs
func(*listargs,**dictkwargs) 
```

**Nested Function**

```python
#Calling the function inside a function
def functionA():
    msg = "Lynda"
    def InnerFA():
        print("Nested Function called")
    return InnerFA() #Executing the InnterFA Function
a = functionA() 
a
```

```python
#Assigning the function in a function to a variable outside a function
def functionA():
    msg = "Lynda"
    def InnerFA():
        print("Nested Function called")
    return InnerFA #Passing the function to a
a = functionA() #a is assigned the function
```

```python
a #Inner function is assigned 
```

```python
a() #Inner function called
```

```python
#Functions inside functions and how to call them
def Outer(a):
    def Middle(b):
        return b+1
    b = Middle(a) #Calling the function, assigning it to a variable and then returning a variable ----- 1
    return b 
Outer(5)
```

```python
#Functions inside functions and how to call them
def Outer(a):
    def Middle(b):
        def Inner(c):
            return c+1
        return Inner(b) #Calling the inner function from the return of the middle function ----- 2 
    return Middle(a) 
Outer(5)
```

```python
#Passing the same as functions
def Outer(a):
    def Middle(b):
        def Inner(c):
            return a+b+c
        return Inner 
    return Middle
a = Outer(5)
b = a(6)
c = b(7)
c
```

**Decorators**

```python
#Example 1
def make_pretty(func):
    def inner():
        print("I got decorated")
        func()
    return inner() #Calling the function

def ordinary():
    print("I am ordinary")
```

```python
make_pretty(ordinary) 
#There is nothing to return as the inner function is called and not passed as parameter
```

```python
#Example 2
def make_pretty(func):
    def inner():
        print("I got decorated")
        func()
    return inner #Passing the function

def ordinary():
    print("I am ordinary")
```

```python
a = make_pretty(ordinary) #---- Step 1
a #Has the inner function #---- Step 2
```

```python
a() #---- Step 3
```

```python
#Example 3
def make_pretty(func):
    def inner():
        print("I got decorated")
        func()
    return inner #Passing the function

@make_pretty
def ordinary():
    print("I am ordinary")
```

```python
ordinary() #----- Single step
```

```python
#Example 3
def make_pretty(func):
    def inner():
        print("I got decorated")
        func()
    return inner() #This doesnt work if you are executing this directy
```

```python
@make_pretty
def ordinary():
    print("I am ordinary") #It makes a call here itself
```

```python
#Using Decorator is a method to extend the behaviour of a function without actually changing the fucntion
#A decorator function takes the function as an argument and returns an function with enhanced behaviour / functionality
```
