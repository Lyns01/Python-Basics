**Creating Dictionary**

```python
# Method 1 - express it as dictionary itself
# All the keys and values should be with '' except for ints and floats
# Should have a :
NameList = {'name' : "Albert", 'age' : 36, 'country' : 'India'}
NameList
```

```python
# Method 2 - using dict()
# All the keys can be without '' but values should have '' if they are strings
# Should have a =
NameList = dict(name = "Albert", age = 36, country = "India")
NameList
```

Output will be the same in methond 1 and 2 - similar to how you define in 1

```python
#The values can also be defined to be a list
NameList = dict(name = ["Albert","Lynda"], age = [36,34], country = ["Norway","India"])
NameList
```

```python
#Duplicates will be overwritten
thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964,
  "year": 2020
}
thisdict
```

**Iterate through the Key Value Pairs**

```python
#To iterate through the keys
for val in thisdict:
    print(val)
```

```python
#To iterate through the values
for val in thisdict.values():
    print(val)
```

```python
#To iterate through the key value pairs
for val1,val2 in thisdict.items():
    print(val1,val2)
```

**Functions**

```python
#To get the keys
keys = thisdict.keys()
keys
```

```python
#To get the values
values = thisdict.values()
values
```

```python
#To add a key value pair
thisdict['Eng'] = 'Lyn'
thisdict
```
