```python
class Car:
    def __init__(self,carwindow,cardoor,seat) : #Dummy variables like the ones we use for functions
        self.window = carwindow
        self.door = cardoor
        self.seat = seat
        
    def self_driving(self,a): #All functions you define inside a class should have "self" as the first argument
        if a > 5:
            print("This is a self driving car")
        else:
            print("This is not a self driving car")
```

```python
car_1 = Car(5,6,7)
car_1.window = 7
```

```python
car_1.window
car_1.seat
car_1.seat = 5
car_1.seat
```

```python
car_1.self_driving(2)
```

```python
class Company(Car):
    def __init__(self,carwindow,cardoor,seat,sensor):
        super().__init__(carwindow, cardoor, seat)
        self.sen = sensor
        
    def camera(self, b):
        if b == True:
            print("Yes, Cam is present for the parking")
        else:
            print("No, it is not available")
```

```python
car_2 = Company(4,5,6,2)
car_2.camera(2)
car_2.self_driving(5)
print(car_2.sen)
car_2.door
```

```python
# Parent class
class Animal:
    def __init__(self, name):
        self.name = name  # Initialize the name of the animal

    def speak(self):
        return f"{self.name} makes a sound"

# Child class inheriting from Animal
class Dog(Animal):
    def __init__(self, name, breed):
        # Call the parent class constructor to initialize name
        super().__init__(name)
        self.breed = breed  # Initialize the breed for Dog
    
    # Overriding the speak method
    def speak(self):
        return f"{self.name} barks"

# Child class inheriting from Animal
class Cat(Animal):
    def __init__(self, name, color):
        # Call the parent class constructor to initialize name
        super().__init__(name)
        self.color = color  # Initialize the color for Cat
    
    # Overriding the speak method
    def speak(self):
        return f"{self.name} meows"

# Create instances of Dog and Cat
dog = Dog("Buddy", "Golden Retriever")
cat = Cat("Whiskers", "Gray")

# Call the speak method
print(dog.speak())  # Output: Buddy barks
print(cat.speak())  # Output: Whiskers meows
```

```python
import requests

# Make a GET request to a URL
response = requests.get("https://jsonplaceholder.typicode.com/posts/1")

# Check if the request was successful (status code 200)
if response.status_code == 200:
    # Print the response data (usually in JSON format)
    data = response.json()  # Parse JSON data from the response
    print(data)
else:
    print("Failed to retrieve data:", response.status_code)
```
