**MAP Functions**

```python
def odd_even(num):
    if num%2 == 0:
        return "Even"
    else:
        return "Odd"

odd_even(6)
```

```python
#When you want to pass a list to the function
#Method 1
lista = [1,2,3,4,5]
for x in lista:
    print(odd_even(x))
```

```python
#Method 2 Map function
#General syntax : map(function,iterable value)
map(odd_even,lista)
```

```python
#Store the values in a list
listb = list(map(odd_even,lista))
listb
```

**Filter Function**

```python
#Similar to the Map function but returns the initial values for which the condition is true
def odd_even(num):
    if num%2 == 0:
        return True
    else:
        return False

lista = [1,2,3,4,5]
listb = list(filter(odd_even, lista))
listb
```

```python
#Pass a lambda function to a filter
lista = [1,2,3,4,5]

a = list(filter(lambda a:a%2==0,lista))
a
```

**Exception Handling**

```python
#General Syntax :
try:
    ##Code where the exception can occur
    a00 = b00
except :
    #The error message you want to print
    print("Give a value for b00 you fool")
```

```python
try:
    a00 = b00
#Prints the actual error
except Exception as ex:
    print(ex) 
```

```python
try:
    a00 = b00
except NameError as ex1: #Possible error 1
    print('User did not define b00') 
except DataError as ex1: #Possible error 2
    print('Datatype din match') 
except Exception as ex: #Any error other than 1 and 2
    print(ex) 
```

There are different types of errors - if you know the type of error - you can print different statements for different errors
Every error has a name - zerodivisionerror, nameerror, typeerror, etc - You can get it from the error description

```python
#Else part - When try part does not throw any error, else part gets excecuted
b00 = 0
try:
    a00 = b00
except Exception as ex: 
    print(ex) 
else :
    print("Executed without error")
```

```python
#Finally - will be executed whatever the situation is - without error
b00 = 0
try:
    a00 = b00
except Exception as ex:
    print(ex) 
else :
    print("Executed without error")
finally:
    print("Done")
```

```python
#Finally - will be executed whatever the situation is - with error
try:
    a00 = b01
except Exception as ex: 
    print(ex) 
else :
    print("Executed without error")
finally:
    print("Done")
```

```python
#Creating a custom exception 
class dobexception(Exception): #Pass the exception raised as an attribute to the class
    pass

year = 2000
age = 2019 - year
try :
    if age>20:
        print("age is valid")
    else:
        raise dobexception #Calls object creation where the parameter is an exception -> exception raise -> except command will be executed
        
except dobexception:
    print("Age is not valid")
```

```python
#Assertion error
a=10
try :
    assert a<10 #Raise an error when the condition is false
except AssertionError as ex: #print(ex) will print nothing because its a custom error
    print("Did not match")
```

**Generator**

```python
#Function with return
def square(n):
    for i in range(n):
        return i**2
#Function with yield
def square1(n):
    for i in range(n):
        yield i**2   

square(3) #0 Because it returns the value of i**2 which is 0**2 = 0
```

```python
square1(3) #appends each value of the loop and creates a generator object - convert that into a list
```

```python
list(square1(3)) 
```
